```python
import networkx as nx
from graph_analysis import analyze_graph_df, visualize_graph_analysis
```

```python
G = nx.Graph()

edges = [
    # --- Cluster A (nodes 1–5, dense) ---
    (1,2), (1,3), (1,4), (1,5),
    (2,3), (2,4), (3,4), (3,5),

    # --- Cluster B (nodes 6–10, dense) ---
    (6,7), (6,8), (6,9), (6,10),
    (7,8), (7,9), (8,9), (9,10),

    # --- Bridge structure ---
    (5,11), (11,6),

    # --- Extra connections to make it interesting ---
    (4,6),   # cross-cluster shortcut
    (3,7),
    (10,12),
    (12,8)
]

G.add_edges_from(edges)
```

```python
results = analyze_graph_df(G, name="Rightmost graph")

for key, value in results.items():
    print(f"\n=== {key} ===")
    display(value)

visualize_graph_analysis(G, results, name="Rightmost graph")
```
